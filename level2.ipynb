# TASK-1   Table Booking and Online Delivery


```python
import pandas as pd

df = pd.read_csv("Dataset .csv")

print(df.head())
```

```python
# Percentage analysis of categorical columns
# Table booking percentage
table_booking = df['Has Table booking'].value_counts(normalize=True) * 100

print("Table Booking Percentage:\n")
print(table_booking)

# Online delivery percentage
online_delivery = df['Has Online delivery'].value_counts(normalize=True) * 100

print("\nOnline Delivery Percentage:\n")
print(online_delivery)
```

```python
# VISUALIZATION
import matplotlib.pyplot as plt
# Table booking percentage
plt.figure(figsize=(6,4))   
plt.bar(table_booking.index, table_booking.values, color=['blue', 'orange'])
plt.title("Table Booking Percentage")
plt.xlabel("Has Table Booking")
plt.ylabel("Percentage")
plt.xticks(rotation=0)
plt.show()
```

```python
# AVERAGE RATING COMPARISON
# Average rating for restaurants with table booking
avg_rating_table_booking = df.groupby('Has Table booking')['Aggregate rating'].mean()
print("\nAverage Rating by Table Booking:\n")
print(avg_rating_table_booking)
# Average rating for restaurants with online delivery
avg_rating_online_delivery = df.groupby('Has Online delivery')['Aggregate rating'].mean()
print("\nAverage Rating by Online Delivery:\n")
print(avg_rating_online_delivery)
```

```python
# Visualization of average ratings by table booking
plt.figure(figsize=(6,4))
avg_rating_table_booking.plot(kind='bar', color=['blue', 'orange'])
plt.title("Average Rating by Table Booking")
plt.xlabel("Has Table Booking")
plt.ylabel("Average Rating")
plt.xticks(rotation=0)
plt.show()
```

```python
# ONLINE DELIVERY VS PRICE RANGE
avg_price_online_delivery = df.groupby('Has Online delivery')['Price range'].mean()
print("\nAverage Price Range by Online Delivery:\n")
print(avg_price_online_delivery)
```

```python
# VISUALIZATION OF AVERAGE PRICE RANGE BY ONLINE DELIVERY
plt.figure(figsize=(6,4))
avg_price_online_delivery.plot(kind='bar', color=['green', 'red'])
plt.title("Average Price Range by Online Delivery")
plt.xlabel("Has Online Delivery")
plt.ylabel("Average Price Range")
plt.xticks(rotation=0)
plt.show()
```

# TASK-3 Feature Engineering

```python
# Restaurant name & address length

# Length of restaurant name
df['Restaurant Name Length'] = df['Restaurant Name'].apply(len)
print(df[['Restaurant Name', 'Restaurant Name Length']].head())

# Length of address length 
df['Address length'] = df['Address'].apply(len)
print(df[['Address', 'Address length']].head()) 
```

```python
# encode table booking
df['Has Table Booking'] = df['Has Table booking'].map({
    'Yes': 1,
    'No': 0
})

print(df[['Has Table booking', 'Has Table Booking']].head()) 
```

```python
# encode delivery online 
df['Has Online Delivery'] = df['Has Online delivery'].map({
    'Yes': 1,
    'No': 0
})

print(df[['Has Online delivery', 'Has Online Delivery']].head())
```
