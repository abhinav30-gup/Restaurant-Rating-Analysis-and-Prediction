# TASK-3 Geospatial Analysis

```python
import folium
```

```python
# create restaurant map
# Create base map
restaurant_map = folium.Map(
    location=[df['Latitude'].mean(), df['Longitude'].mean()],
    zoom_start=5
)

# Add markers
for i in range(100):
    folium.Marker(
        [df.iloc[i]['Latitude'], df.iloc[i]['Longitude']],
        popup=df.iloc[i]['Restaurant Name']
    ).add_to(restaurant_map)

# Display map
restaurant_map
```

```python
# Restaurant by city and countries
city_distribution = df['City'].value_counts().head(10)
country_distribution = df['Country Code'].value_counts()
print(city_distribution)
print(country_distribution)
```

```python
# visualization
import matplotlib.pyplot as plt
plt.figure(figsize=(6,4))
plt.bar(city_distribution.index, city_distribution.values)
plt.title("Top 10 Cities with Highest Restaurants")
plt.xlabel("City")
plt.ylabel("Number of Restaurants")
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
# visualization
import matplotlib.pyplot as plt
plt.figure(figsize=(6,4))
plt.bar(country_distribution.index, country_distribution.values)
plt.title("Top 10 Countries with Highest Restaurants")
plt.xlabel("Country Code")
plt.ylabel("Number of Restaurants")
plt.tight_layout()
plt.show()
```

```python
# Location vs rating correlation
correlation = df[['Latitude', 'Longitude', 'Aggregate rating']].corr()

print(correlation)
```

```python
# visualization
import seaborn as sns
sns.heatmap(correlation, annot=True)
plt.title("Correlation Between Location and Rating")
plt.show()
```
